```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedKFold
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report
from sklearn import preprocessing
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.model_selection import learning_curve
from sklearn.preprocessing import normalize
from sklearn.preprocessing import LabelEncoder
```

```python
df_cancer = pd.read_csv("breast_cancer.csv")

print(df_cancer.info())
print(df_cancer.describe())
print("Number of class of the target : " ,df_cancer['Class'].unique().size)
```

```python
print(df_cancer.head())
print(df_cancer.dtypes)
```

```python
print(df_cancer["Bare_Nuclei"].unique())
df_cancer["Bare_Nuclei"] = pd.to_numeric(df_cancer["Bare_Nuclei"], errors='coerce')
print(df_cancer["Bare_Nuclei"].unique())
print(df_cancer["Bare_Nuclei"].dtype)
```

```python
df_cancer = df_cancer.dropna(subset="Bare_Nuclei")
print(df_cancer.describe())
print(df_cancer["Bare_Nuclei"].unique())
```

```python
df_cancer["Class"] = df_cancer["Class"].map( lambda x : 0 if x == 2 else 1)
print(df_cancer["Class"].unique())
```

```python
y = df_cancer["Class"]
X = df_cancer.drop(columns=["Class", "Sample_code_number"])
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.33, random_state=42)
```

```python
lf = SVC(kernel='linear')
lf.fit(X_train, y_train)
print("Linear : ",lf.score(X_test, y_test))
```

```python
lf_rbf = SVC(kernel='rbf', max_iter=10000)
lf_rbf.fit(X_train, y_train)
print("RBF : ",lf_rbf.score(X_test, y_test))
```

```python
X_l1 = normalize(X_train, norm='l1')
X_test_l1 = normalize(X_test, norm='l1')

lf_lin_l1 = SVC(kernel='linear', random_state=42, max_iter=10000)
lf_lin_l1.fit(X_l1, y_train)
print("Linear l1 : ",lf_lin_l1.score(X_test_l1, y_test))
lf_rbf_l1 = SVC(kernel='rbf', max_iter=10000)
lf_rbf_l1.fit(X_l1, y_train)
print("RBF l1 : ",lf_rbf_l1.score(X_test_l1, y_test))


X_l2 = normalize(X_train, norm='l2')
X_test_l2 = normalize(X_test, norm='l2')

lf_lin_l2 = SVC(kernel='linear', max_iter=10000)
lf_lin_l2.fit(X_l2, y_train)
print("Linear l2 : ", lf_lin_l2.score(X_test_l2, y_test))
lf_rbf_l2 = SVC(kernel='rbf', max_iter=10000)
lf_rbf_l2.fit(X_l2, y_train)
print("RBF l2 : ",lf_rbf_l2.score(X_test_l2, y_test))
```

```python
y_pred = lf_rbf.predict(X_test)
confusion = confusion_matrix(y_test, y_pred)
print(confusion)
print(classification_report(y_test, y_pred))
```

```python
def select_k_best(X, y, k):
    """Select k best features using chi-squared test"""
    selector = SelectKBest(chi2, k=k)
    X_new = selector.fit_transform(X, y)
    
    # Get selected feature indices
    mask = selector.get_support()
    feature_names = X.columns if hasattr(X, 'columns') else range(X.shape[1])
    selected_features = [f for f, m in zip(feature_names, mask) if m]
    
    print(f"Selected {k} best features:")
    print(selected_features)
    print(f"Feature scores: {selector.scores_[mask]}")
    
    return X_new, selected_features
```

```python
X_5_best, selected_features = select_k_best(X, y, 5)
```

```python
linear_svc_5 = SVC(kernel='linear', random_state=42)
linear_scores_5 = cross_val_score(linear_svc_5, X_5_best, y, cv=3)
print("Linear SVC Linear (5 features) scores:", linear_scores_5)

linear_svc_5 = SVC(kernel='rbf', random_state=42)
linear_scores_5 = cross_val_score(linear_svc_5, X_5_best, y, cv=3)
print("Linear SVC RBF (5 features) scores:", linear_scores_5)
```

```python
parameters = {
    'kernel': ['linear', 'rbf', 'poly'],
    'C': [0.1, 1, 10, 100],
    'gamma': ['scale', 'auto', 0.001, 0.01, 0.1, 1]
}

svc = SVC(random_state=42)
grid_search = GridSearchCV(svc, parameters, cv=3, scoring='accuracy', n_jobs=-1, verbose=1)
grid_search.fit(X, y)

print("\nBest parameters:", grid_search.best_params_)
print("Best score: {:.4f}".format(grid_search.best_score_))
#So, yes
```

```python
def do_cv(model, X, y, cv=3):
    scores = cross_val_score(model, X, y, cv=cv)
    print(f"Model: {model}")
    print(f"Mean: {scores.mean():.4f}")
    print(f"Std: {scores.std():.4f}")
    return scores.mean(), scores.std()
```

```python
def do_cm_cr(model, X, y, names):
    X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, test_size=0.33, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred, target_names=names))

```

```python
def do_grid_search(model, parameters, X, y, cv=3):
    grid_search = GridSearchCV(model, parameters, cv=cv, scoring='accuracy', n_jobs=-1, verbose=1)
    grid_search.fit(X, y)
    print(grid_search.best_params_)
    print(grid_search.best_score_)
    return grid_search.best_estimator_
```
